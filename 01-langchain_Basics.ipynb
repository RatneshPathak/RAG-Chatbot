# 01 · LangChain basics with a local Ollama model

Calls a local LLM through `ChatOllama`.

**Prerequisites** (see the README for details)

1. Install [Ollama](https://ollama.com) and make sure it is running (`ollama serve`, or the background service on Windows/macOS).
2. Download the model(s) used here:

```powershell
ollama pull gemma3:1b
```
3. `pip install -r requirements.txt` inside a Python 3.12 virtual environment, and select that environment as this notebook's kernel.

Run cells top to bottom. *Author: Ratnesh Pathak*

> **About the saved outputs:** outputs shown below were produced during earlier local runs and were *not* re-executed when this repository was audited (no Ollama server was available). Model replies vary from run to run.

````python
from langchain_ollama import ChatOllama

base_url = "http://localhost:11434" #this is where Ollama is running
model = "gemma3:1b"  # specify the model you want to use

llm = ChatOllama(
    base_url=base_url,
    model=model,
    temperature=0.7,
    num_predict=200)  # cap on generated tokens (ChatOllama has no `max_tokens` option)
````

````python
response = llm.invoke("What is the capital of France?")
print(response.content)
````

````python
response = llm.invoke("Where is Thakur College of science and commerce located in mumbai?")
print(response.content)
````

> ⚠️ **Why RAG matters:** a small model answers confidently about a place it does not actually know and can get the details wrong. The repository's own note, `data/thakur_college_info.txt`, gives the real location. Retrieval-augmented generation (the later notebooks) supplies such facts to the model instead of relying on its memory.
